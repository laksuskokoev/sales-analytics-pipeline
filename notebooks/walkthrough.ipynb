# Sales analytics — walkthrough
Synthetic dataset. The analysis uses reproducible order-level metrics and reconciles SQL aggregates.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd()
if not (root / "reports" / "summary.csv").exists():
    root = root.parent
summary = pd.read_csv(root / "reports" / "summary.csv")
summary
```

## Definitions
Only `completed` orders count toward net_sales and product_contribution. Product contribution excludes non-product costs.

```python
categories = pd.read_csv(root / "reports" / "categories.csv")
channels = pd.read_csv(root / "reports" / "channels.csv")
categories.assign(contribution_margin_pct=(100 * categories.product_contribution / categories.net_sales).round(2))
```

## Reconciliation: category and channel breakdowns must match overall sales.

```python
assert abs(categories.net_sales.sum() - summary.net_sales.iloc[0]) < 0.01
assert abs(channels.net_sales.sum() - summary.net_sales.iloc[0]) < 0.01
print("OK")
```

## Limitations
- No advertising costs, so cannot compute channel ROI.
- No visits/sessions, so completion rate is not visitor conversion.
- Simplified full-return treatment; this is not accounting revenue or net profit.
